# MNIST: convolutional neural networks with PyTorch

Convolutional neural networks (CNNs) exploit spatial structure in images. In this notebook we train a simple CNN to recognize handwritten digits, then investigate how an additional dense layer and dropout affect its performance.

## Configuration

The full experiment trains each model for ten epochs. The default quick configuration uses three epochs so that the complete workflow remains practical on a CPU; set `FULL_EXPERIMENT = True` for the full comparison.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader, random_split
from torchvision.datasets import MNIST
from torchvision.transforms import v2
```

```python
FULL_EXPERIMENT = False

BATCH_SIZE = 32
EPOCHS = 20 if FULL_EXPERIMENT else 3

SPLIT_SEED = 1234
MODEL_SEED = 40_580
DATA_ORDER_SEED = 2718

DATA_DIR = Path.cwd() / "data"
MODEL_DIR = Path.cwd() / "models"
```

## Reproducibility and device selection

A fixed seed makes a CPU run reproducible under the same software and hardware configuration. Exact results can still differ across devices or PyTorch versions because some accelerator operations are nondeterministic or use different numerical implementations.


```python
def seed_everything(seed):
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


seed_everything(MODEL_SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
using_cuda = device.type == "cuda"
print(f"Using device: {device}")
```

## Data preparation

### Obtaining and transforming the dataset

MNIST contains grayscale images with 28 × 28 pixels. PyTorch represents an image as `[channels, height, width]`, so each transformed example has shape `[1, 28, 28]`.

The transformation converts each image to a `float32` tensor and scales its pixel values from the integer range 0–255 to the floating-point range 0–1.


```python
transform = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
])
```

```python
full_train_dataset = MNIST(
    root=DATA_DIR,
    train=True,
    download=True,
    transform=transform,
)

test_dataset = MNIST(
    root=DATA_DIR,
    train=False,
    download=True,
    transform=transform,
)
```

### Validation set

75% of the original training set is used for training and 25% for validation. The test set remains untouched while architectures and training choices are compared.

```python
split_generator = torch.Generator().manual_seed(SPLIT_SEED)

train_dataset, validation_dataset = random_split(
    full_train_dataset,
    [45_000, 15_000],
    generator=split_generator,
)

len(train_dataset), len(validation_dataset), len(test_dataset)
```

### Data loaders

The training loader shuffles examples at every epoch. Recreating it with the same generator seed gives both candidate models the same sequence of mini-batches. Validation and test loaders do not shuffle. When CUDA is used, pinned CPU memory and non-blocking device transfers can work together to make host-to-device copies asynchronous.


```python
def make_train_loader(seed=DATA_ORDER_SEED):
    generator = torch.Generator().manual_seed(seed)
    return DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        generator=generator,
        num_workers=0,
        pin_memory=using_cuda,
    )


training_evaluation_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=using_cuda,
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=using_cuda,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=using_cuda,
)

development_loaders = {
    "train": training_evaluation_loader,
    "validation": validation_loader,
}
```

### Loss function and optimizer

The optimizer uses stochastic gradient descent with learning rate 0.1, momentum, Nesterov acceleration, and a small weight-decay term. Exact optimizer details can differ between frameworks, so matching named hyperparameters does not guarantee bitwise-identical training.

```python
loss_function = nn.CrossEntropyLoss()
simple_optimizer = torch.optim.SGD(
    simple_model.parameters(),
    lr=0.1,
    momentum=0.9,
    weight_decay=1.0e-6,
    nesterov=True,
)
```

### Training and evaluation functions

The training loop switches the model to training mode, computes logits and loss, propagates gradients, updates the parameters, and aggregates metrics over all examples. Evaluation switches to evaluation mode—essential when dropout is present—and disables gradient tracking.


```python
def train_one_epoch(
    model,
    data_loader,
    loss_function,
    optimizer,
    device,
    non_blocking=False,
):
    model.train()
    total_loss = 0.0
    total_correct = 0
    total_examples = 0

    for features, targets in data_loader:
        features = features.to(device, non_blocking=non_blocking)
        targets = targets.to(device, non_blocking=non_blocking)

        optimizer.zero_grad(set_to_none=True)
        logits = model(features)
        loss = loss_function(logits, targets)
        loss.backward()
        optimizer.step()

        batch_size = targets.size(0)
        total_loss += loss.item() * batch_size
        total_correct += (logits.argmax(dim=1) == targets).sum().item()
        total_examples += batch_size

    return {
        "loss": total_loss / total_examples,
        "accuracy": total_correct / total_examples,
    }
```

```python
@torch.inference_mode()
def evaluate(
    model,
    data_loader,
    loss_function,
    device,
    non_blocking=False,
):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total_examples = 0

    for features, targets in data_loader:
        features = features.to(device, non_blocking=non_blocking)
        targets = targets.to(device, non_blocking=non_blocking)

        logits = model(features)
        loss = loss_function(logits, targets)

        batch_size = targets.size(0)
        total_loss += loss.item() * batch_size
        total_correct += (logits.argmax(dim=1) == targets).sum().item()
        total_examples += batch_size

    return {
        "loss": total_loss / total_examples,
        "accuracy": total_correct / total_examples,
    }
```

```python
def fit(
    model,
    train_loader,
    validation_loader,
    loss_function,
    optimizer,
    epochs,
    device,
    non_blocking=False,
):
    history = {
        "loss": [],
        "accuracy": [],
        "val_loss": [],
        "val_accuracy": [],
    }

    for epoch in range(1, epochs + 1):
        train_metrics = train_one_epoch(
            model,
            train_loader,
            loss_function,
            optimizer,
            device,
            non_blocking,
        )
        validation_metrics = evaluate(
            model,
            validation_loader,
            loss_function,
            device,
            non_blocking,
        )

        history["loss"].append(train_metrics["loss"])
        history["accuracy"].append(train_metrics["accuracy"])
        history["val_loss"].append(validation_metrics["loss"])
        history["val_accuracy"].append(
            validation_metrics["accuracy"]
        )

        print(
            f"Epoch {epoch:2d}/{epochs}: "
            f"loss={train_metrics['loss']:.4f}, "
            f"accuracy={train_metrics['accuracy']:.4f}, "
            f"val_loss={validation_metrics['loss']:.4f}, "
            f"val_accuracy={validation_metrics['accuracy']:.4f}"
        )

    return history
```

### Training the simple model


```python
simple_train_loader = make_train_loader()

simple_history = fit(
    model=simple_model,
    train_loader=simple_train_loader,
    validation_loader=validation_loader,
    loss_function=loss_function,
    optimizer=simple_optimizer,
    epochs=EPOCHS,
    device=device,
    non_blocking=using_cuda,
)
```

Plot training and validation metrics together. A widening gap can indicate overfitting, but the curves must be interpreted alongside their absolute values and whether validation performance has stopped improving.


```python
def plot_history(history):
    epochs = np.arange(1, len(history["loss"]) + 1)
    figure, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(epochs, history["loss"], label="Training")
    axes[0].plot(epochs, history["val_loss"], label="Validation")
    axes[0].set(xlabel="Epoch", ylabel="Loss")
    axes[0].legend()

    axes[1].plot(epochs, history["accuracy"], label="Training")
    axes[1].plot(epochs, history["val_accuracy"], label="Validation")
    axes[1].set(xlabel="Epoch", ylabel="Accuracy")
    axes[1].legend(loc="lower right")

    figure.tight_layout()


plot_history(simple_history)
```

Compare performance on the training and validation sets. These development results—not the test set—are used to diagnose and select models.


```python
def evaluate_development_sets(
    model,
    data_loaders,
    loss_function,
    device,
    non_blocking=False,
):
    results = {}
    for name, data_loader in data_loaders.items():
        results[name] = evaluate(
            model,
            data_loader,
            loss_function,
            device,
            non_blocking,
        )
        metrics = results[name]
        print(
            f"{name:10s}: loss={metrics['loss']:.4f}, "
            f"accuracy={metrics['accuracy']:.4f}"
        )

    return results


simple_results = evaluate_development_sets(
    simple_model,
    development_loaders,
    loss_function,
    device,
    non_blocking=using_cuda,
)
```
